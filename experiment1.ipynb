```python
import numpy as np
import pandas as pd

# student marks
marks = np.array([78, 85, 92, 67, 88, 74])

print("Student marks:", marks)

print("Mean =", np.mean(marks))
print("Median =", np.median(marks))
print("Maximum =", np.max(marks))
print("Minimum =", np.min(marks))
print("Total marks =", np.sum(marks))
print("Standard deviation =", np.std(marks))

print("\nAfter adding 5:", marks + 5)
print("After subtracting 5:", marks - 5)

print("\nFirst three marks:", marks[:3])
print("Last three marks:", marks[-3:])
print("Marks from index 1 to 4:", marks[1:5])


students = {
    "Name": ["Aman", "Rahul", "Priya", "Sneha", "Arjun"],
    "Marks": [78, 85, 92, 67, 88],
    "Branch": ["CSE", "CSE", "ISE", "CSE", "ISE"]
}

df = pd.DataFrame(students)

print("\nStudent details:")
print(df)

print("\nFirst three students:")
print(df.head(3))

print("\nStudent names:")
print(df["Name"])

print("\nStudents scoring more than 80:")
print(df[df["Marks"] > 80])

print("\nFirst two rows:")
print(df.iloc[:2])
```
